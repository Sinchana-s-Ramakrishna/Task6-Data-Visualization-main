```python
import pandas as pd
df = pd.read_csv("World Happiness Report 2005-2021.csv")
df.head()
```

```python
df.columns
df.info()
```

```python
import matplotlib.pyplot as plt

top10 = df.sort_values(by="Life Ladder", ascending=False).head(10)

plt.figure(figsize=(10,6))
plt.bar(top10["Country_name"], top10["Life Ladder"])
plt.title("Top 10 Happiest Countries")
plt.xlabel("Country")
plt.ylabel("Happiness Score")
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.show()

```

```python
plt.figure(figsize=(10,6))
plt.plot(top10["Country_name"], top10["Life Ladder"], marker='o')
plt.title("Happiness Score Trend Among Top 10 Countries")
plt.xlabel("Country")
plt.ylabel("Happiness Score")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Life Ladder"], bins=15)
plt.title("Distribution of Happiness Scores")
plt.xlabel("Happiness Score")
plt.ylabel("Number of Countries")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Log GDP per capita"], df["Life Ladder"])
plt.title("GDP vs Happiness Score")
plt.xlabel("Log GDP per Capita")
plt.ylabel("Happiness Score")
plt.grid(True)
plt.show()
```

Insight 1: Top happiest countries have consistently high life satisfaction levels.

Insight 2: Global happiness trend shows fluctuations that may align with major world events.

Insight 3: There is a positive correlation between GDP per capita and happiness score.
